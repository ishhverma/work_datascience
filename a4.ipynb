```python
from typing import List

class Solution:
    def maxProduct(self, nums: List[int]) -> int:
        nums.sort()
        return (nums[-1] - 1) * (nums[-2] - 1)
```

```python
from typing import List

class Solution:
    def numTeams(self, rating: List[int]) -> int:
        n = len(rating)
        ans = 0

        for j in range(n):
            left_smaller = left_larger = 0
            right_smaller = right_larger = 0

            # left side
            for i in range(j):
                if rating[i] < rating[j]:
                    left_smaller += 1
                else:
                    left_larger += 1

            # right side
            for k in range(j + 1, n):
                if rating[k] > rating[j]:
                    right_larger += 1
                else:
                    right_smaller += 1

            ans += left_smaller * right_larger
            ans += left_larger * right_smaller

        return ans
```

```python
class Solution:
    def numberOfSteps(self, num: int) -> int:
        steps = 0

        while num > 0:
            if num % 2 == 0:
                num //= 2   # even
            else:
                num -= 1    # odd
            steps += 1

        return steps
```

```python
from typing import List

class Solution:
    def countBits(self, n: int) -> List[int]:
        ans = [0] * (n + 1)

        for i in range(1, n + 1):
            ans[i] = ans[i // 2] + (i % 2)

        return ans
```
