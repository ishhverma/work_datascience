```python
from typing import List

class Solution:
    def findNumbers(self, nums: List[int]) -> int:
        count = 0

        for num in nums:
            if len(str(num)) % 2 == 0:
                count += 1

        return count
```

```python
from typing import List

class Solution:
    def numIdenticalPairs(self, nums: List[int]) -> int:
        freq = {}
        count = 0

        for num in nums:
            if num in freq:
                count += freq[num]
                freq[num] += 1
            else:
                freq[num] = 1

        return count
```

```python
class Solution:
    def subtractProductAndSum(self, n: int) -> int:
        product = 1
        sum_digits = 0

        while n > 0:
            digit = n % 10
            product *= digit
            sum_digits += digit
            n //= 10

        return product - sum_digits
```

```python
class Solution:
    def xorOperation(self, n: int, start: int) -> int:
        result = 0

        for i in range(n):
            result ^= start + 2 * i

        return result
```
